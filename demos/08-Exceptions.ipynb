## Captures d'excpetions

```python
user_value = "dix"

def convert(data):
    value = int(data)
    return value

print("start")

try:
    value = convert(user_value)
    print(value + 10)

except IndexError:
    print('autre chose')
except (ValueError, LookupError):
    print('Capturée')
    


print("end")
```

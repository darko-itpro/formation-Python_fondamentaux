## Exercice

```python
from datetime import datetime, timedelta
import pyflix.datasource as ds
```

```python
start_hours, start_minutes = ds.get_start_time().split('h')
start_hours = int(start_hours)
start_minutes = int(start_minutes)
```

```python
tonight = datetime.now().replace(hour=start_hours, minute=start_minutes, # <- obligatoire
                                 second=0, microsecond=0)                # <- parce que c'est bien
```

```python
n_episodes = 10
EPISODE_DURATION = 23

view_duration = timedelta(minutes= EPISODE_DURATION * n_episodes)
```

```python
end_time = tonight + view_duration
```

```python
end_time
```

```python
f"Dodo le {end_time:%d/%m} {end_time:%H:%M}"
```

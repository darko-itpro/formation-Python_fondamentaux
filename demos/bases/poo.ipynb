## Déclaration de méthodes et attributs

```python
class Training:
    def __init__(self, subject:str, duration:int, seats:int=12):
        self.subject = subject
        self.duration = duration
        self.seats = seats
        self.students: list[str] = []

    def add_student(self, name:str):
        self.students.append(name)


t = Training('Python', 5)
```

## Attention au name mangling

```python
class Training:
    def __init__(self, subject:str, duration:int, seats:int=12):
        self.subject = subject
        self.duration = duration
        self.seats = seats
        self.__students: list[str] = []

    def add_student(self, name:str):
        self.__students.append(name)
```

```python
t = Training('Python', 5)
```

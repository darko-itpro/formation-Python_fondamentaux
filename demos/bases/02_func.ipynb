# Les fonctions

```python
def say_hello():
    print("Hello World")

print(say_hello())
```

```python
def give_hello() -> str:
    value = "Hello given"
    return value

print(give_hello())
```

```python
def say_hello_to(name:str):
    print("Hello", name)

say_hello_to("Guido")
```

## Paramètres optionnels

```python
def say_hello(name:str="World"):
    print("Hello", name)
```

```python
say_hello()
say_hello("Guido")
```

```python
def create_account(acc_id:str, value:float=100, overdraft:bool=False):
    print(acc_id, value, overdraft)
```

```python
create_account("acc Id", 100.0, True)
```

```python
create_account("acc Id", 100.0)
create_account("acc Id")
```

```python
create_account(overdraft=True, acc_id="acc Id")
```

```python
def create_account(acc_id:str, /, value:float=100, overdraft:bool=False):
    print(acc_id, value, overdraft)
```

```python
create_account(acc_id="acc Id", value=20000, overdraft=True)
```

```python
def create_account(acc_id:str, value:float=100, *, overdraft:bool=False):
    print(acc_id, value, overdraft)
```

```python
create_account("mui", 1555, True)
```

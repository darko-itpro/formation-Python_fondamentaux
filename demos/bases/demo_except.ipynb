## Capture

```python
data = "dix"

def transform(value):
    int(value)

try:
    answer = transform(data)
    print(f"Answer is {answer}")
except IndexError:
    answer = None
except (ValueError, LookupError):
    print("oups erreur")
finally:
    print("cleanup")



print("done")
```

## Capture des exceptions

```python
value = "dix"

def convert(data):
    print('before')
    v = int(data)
    print('after')
    return v

try:
    result = convert(value)
    print(f"after is {result}")
except KeyError:
    print("Autre grestion de l'erreur")
except (ValueError, LookupError):
    print('Value ignored bikose erreur mesure') # Normalement un log


print("End")
```

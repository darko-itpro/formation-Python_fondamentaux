### Seconde exercice

```python
import pyflix.datasource as ds

episodes = 7
EPISODE_DURATION = 23

show_duration = episodes * EPISODE_DURATION  # En minutes

hours, minutes = ds.get_start_time().split("h")
hours = int(hours)
minutes = int(minutes)

end_time = hours * 60  + minutes + show_duration

"J'irai au lit à {}h{:02}".format(end_time // 60, end_time % 60)
```

```python
"J'irai au lit à {}h{:02}".format(*divmod(end_time, 60))
```

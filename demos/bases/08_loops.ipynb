# Les boucles

## Le `for`

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
for knight in camelot:
    print(knight.upper())
```

```python
for index, knight in enumerate(camelot):
    if index % 2:
        print(">", knight)
    else:
        print("-", knight)
```

```python
for _ in range(4):
    print("turn left")
```

## Itération avec `while`

```python
value = 1
previous = 1

while value < 5000:
    print(value/previous)
    value, previous = value + previous, value
```

```python
camelot = [("Arthur", "king"), ("Merlin", "wizard"),
           ("Lancelot", "knight"), ("Robin", "knight"),
           ("Perceval", "knight"), ("Karadoc", "knight")]
```

```python
for knight, job in camelot:
    print("Hail", knight)
    if job == "wizard":
        continue
    print("He is a mighty knight")
```

```python
for knight, job in camelot:
    print(knight)
    if job == "wizard":
        print("Good, you have a wizard")
        break
```

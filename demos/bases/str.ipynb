## Formatage de chaines

```python
template_fr = "Prix : {:10.2f} {}"

print(template_fr.format(42, '€'))
print(template_fr.format(42.678, '€'))
print(template_fr.format(576554566, '€'))
```

```python
template_us = "Price : {1}{0:.2f}"
print(template_us.format(65, '$'))
```

```python
"Il est {}h{:02}".format(14, 8)
```

```python
price = 42
currency = "€"
```

```python
f"Prix : {price:.2f} {currency}"
```

```python
%timeit template_fr.format(price, currency)
```

```python
%timeit  f"Prix : {price:.2f} {currency}"
```

### Second exercice

```python
import pyflix.datasource as ds
```

```python
EPISODE_DURATION = 23
episodes_to_watch = 7

watch_duration = EPISODE_DURATION * episodes_to_watch
```

```python
hours, minutes = ds.get_start_time().split("h")
hours = int(hours)
minutes = int(minutes)

start_time = hours * 60 + minutes
```

```python
end_time = start_time + watch_duration
```

```python
print(hours, minutes)
print(end_time)
print("J'irai au lit à {:02}h{:02}".format(end_time // 60, end_time % 60))
```

# List Comprehension

```python
camelot = [("Arthur", "king"), ("Merlin", "wizard"),
           ("Lancelot", "knight"), ("Robin", "knight"),
           ("Perceval", "knight"), ("Karadoc", "knight")]
```

```python
[name for name, job in camelot]
```

```python
[len(name) for name, job in camelot]
```

```python
[item for item in camelot if item[1] != "wizard"]
```

```python
[name for name, job in camelot if job != "wizard"]
```

```python
camelot = [("Arthur", "king"), ("Merlin", "wizard"),
           ("Lancelot", "knight"), ("Robin", "knight"),
           ("Perceval", "knight"), ("Karadoc", "knight")]
```

```python
for i in (name for name, job in camelot if job != "wizard"):
    print(i)
```

```python
def get_quest_candidates(kigdom):
    for knight, job in kigdom:
        if job != "wizard":
            yield knight

    print("done !!!")
```

```python
for name in get_quest_candidates(camelot):
    print(name)
```

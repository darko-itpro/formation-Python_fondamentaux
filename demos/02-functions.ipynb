# Les fonctions

```python
def say_hello() -> None:
    print('Hello World')
```

```python
say_hello()
```

# Avec retour

```python
def give_hello() -> str:
    value = "Hello given"
    return value
```

```python
give_hello()
```

# Les paramètres

```python
def say_hello_to(name:str, title:str):
    print("Hello", title, name)
```

```python
say_hello_to("Guido", "Mr.")
```

## Paramètres optionnels

```python
def say_hello(name:str = "World"):
    print("Hello", name)
```

```python
say_hello()
say_hello("Guido")
```

```python
def create_account(acc_id:str, value:int = 100, overdraft:bool = False):
    print(acc_id, value, overdraft)
```

```python
create_account("My id", 100_000, True)
```

```python
create_account("My id", 100_000)
create_account("My id")
```

```python
create_account(acc_id="Miu ud", overdraft=True)
```

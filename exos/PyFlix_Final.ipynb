```python
ASSETS_ROOT = Path('/formation-Python_fondamentaux/assets') # Précisez ici le chemin racine des assets
media_path = ASSETS_ROOT / 'files'
csv_path = ASSETS_ROOT / 'showslist.csv'

if not media_path.exists():
    raise FileNotFoundError("Le chemin vers les fichiers médias ou le fichier CSV n'existe pas.")
```

```python
#import pyflix.mediatheque as md
import pyflix.media_db as md

shows = {}

for episode_info in fu.load_medias_from_csv(csv_path):
    show_name = episode_info[0]
    try:
        show = shows[show_name]
    except KeyError:
        show = md.TvShow(show_name)
        shows[show_name] = show
        
    try:
        show.add_episode(*episode_info[1:])
    except ValueError:
        print(f"Episode {episode_info[1]} du show {show_name} déjà présent.")
```

```python
for episode_info in fu.get_media_info(media_path):
    show_name = episode_info[0]
    try:
        show = shows[show_name]
    except KeyError:
        show = md.TvShow(show_name)
        shows[show_name] = show

    try:
        show.add_episode(*episode_info[1:])
    except ValueError:
        print(f"Episode {episode_info[1]} du show {show_name} déjà présent.")
```

```python
import pyflix.load_and_display as ld
ld.display_all_shows(shows)
```
